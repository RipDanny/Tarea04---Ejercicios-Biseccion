1. Use el método de bisección para encontrar soluciones precisas dentro de $10^{-2}$ para $x^3 -7x^2 +14x -6=0$ en cada intervalo.  
a. [0, 1] 


```python
def f(x):
    return x**3 - 7*x**2 + 14*x - 6

def biseccion(a, b, tol=1e-2, maxiter=100):
    fa = f(a)
    fb = f(b)

    if fa * fb > 0:
        raise ValueError(f"No hay cambio de signo en [{a}, {b}]")

    for i in range(1, maxiter + 1):
        m = (a + b) / 2
        fm = f(m)

        if (b - a) / 2 <= tol:
            return m, i

        if fm == 0:
            return m, i

        if fa * fm < 0:
            b = m
            fb = fm
        else:
            a = m
            fa = fm

    return m, maxiter


raiz_a, it_a = biseccion(0, 1, tol=1e-2)
print("a) Intervalo [0,1]")
print("  Raíz aproximada:", raiz_a)
print("  Iteraciones:", it_a)
print("  f(raíz) =", f(raiz_a))
print()
```

b. [1,3.2]

```python
raiz_b, it_b = biseccion(1, 3.2, tol=1e-2)
print("b) Intervalo [1, 3.2]")
print("  Raíz aproximada:", raiz_b)
print("  Iteraciones:", it_b)
print("  f(raíz) =", f(raiz_b))
print()
```

c. [3.2,4]

```python
raiz_c, it_c = biseccion(3.2, 4, tol=1e-2)
print("c) Intervalo [3.2, 4]")
print("  Raíz aproximada:", raiz_c)
print("  Iteraciones:", it_c)
print("  f(raíz) =", f(raiz_c))
```

**b. Use el método de bisección para encontrar soluciones precisas dentro de $10^-5 para el primer valor positivo 
de 𝑥 con 𝑥 = 2 sin 𝑥.$**

```python
import numpy as np

def f(x):
    return x - 2*np.sin(x)

a = 1
b = 2

tol = 1e-5

while (b - a)/2 > tol:

    c = (a + b)/2

    if f(c) == 0:
        break

    elif f(a)*f(c) < 0:
        b = c
    else:
        a = c

raiz = (a + b)/2

print("La raíz aproximada es:", raiz)
print("f(raíz) =", f(raiz))
```

**b. Use el método de bisección para encontrar una aproximación dentro de $10^-3$ para un valor en [−2, 0] con
$𝑥^2 − 1 = e^{(1-x^2)}$**

```python
import numpy as np

def f(x):
    return x**2 - 1 - np.exp(1 - x**2)

a = -2
b = 0

tol = 1e-3

while (b - a)/2 > tol:

    c = (a + b)/2

    if f(c) == 0:
        break

    elif f(a) * f(c) < 0:
        b = c
    else:
        a = c

raiz = (a + b)/2

print("La raíz aproximada es:", raiz)
print("f(raíz) =", f(raiz))
```

1. Un abrevadero de longitud 𝐿 tiene una sección transversal en forma de semicírculo con radio 𝑟. (Consulte la
figura adjunta.) Cuando se llena con agua hasta una distancia ℎ a partir de la parte superior, el volumen 𝑉 de
agua es

$$ V= L[0.5pi r^2 - r^2 arcsen \frac{h}{r}-h(r^2 - h^2)^{(1/2)}]


```python
import numpy as np

def f(h):
    return (np.pi/2) - np.arcsin(h) - h*np.sqrt(1 - h**2) - 1.24

a = 0
b = 1

tol = 0.01

while (b - a)/2 > tol:

    c = (a + b)/2

    if f(c) == 0:
        break

    elif f(a) * f(c) < 0:
        b = c
    else:
        a = c

h = (a + b)/2

print("La profundidad aproximada es:", h, "cm")
print("f(h) =", f(h))
```

2. Un objeto que cae verticalmente a través del aire está sujeto a una resistencia viscosa, así como a la fuerza
de gravedad. Suponga que un objeto con masa 𝑚 cae desde una altura 𝑠଴ y que la altura del objeto después
de 𝑡 segundos es

$$s(t) = S_0 - \frac{mg}{k}t + \frac{m^2 g }{k^2}(1-e^{(-kt/m)})$$

donde $𝑔 = 9.81 m/s^2$ y 𝑘 representa el coeficiente de la resistencia del aire en $Ns/m$. Suponga $s_0 = 300 𝑚$ ,
𝑚 = 0.25 𝑘𝑔 y 𝑘 = 0.1 𝑁𝑠/𝑚. Encuentre, dentro de 0.01 𝑠𝑒𝑔𝑢𝑛𝑑𝑜𝑠, el tiempo que tarda un cuarto de kg en
golpear el piso.

```python
import numpy as np

def s(t):
    return 300 - 24.525*t + 61.3125*(1 - np.exp(-0.4*t))

a = 10
b = 20

tol = 0.01

while (b - a)/2 > tol:

    c = (a + b)/2

    if s(c) == 0:
        break

    elif s(a) * s(c) < 0:
        b = c
    else:
        a = c

t = (a + b)/2

print("Tiempo aproximado:", t, "segundos")
print("s(t) =", s(t))
```

1. La función definida por 𝑓(𝑥) = sin 𝜋𝑥 tiene ceros en cada entero. Muestre cuando −1 < 𝑎 < 0 y 2 < 𝑏 <
3, el método de bisección converge a:


Sea la función:

$$
f(x)=\sin(\pi x)
$$

Sabemos que:

$$
\sin(\pi x)=0
$$

cuando:

$$
x=n, \quad n\in\mathbb{Z}
$$

Por tanto, la función tiene ceros en todos los enteros:

$$
\dots,-2,-1,0,1,2,3,\dots
$$

Además se tiene que:

$$
-1<a<0
$$

y

$$
2<b<3
$$

Por lo tanto, el intervalo inicial del método de bisección es:

$$
[a,b]
$$

Punto medio inicial

El método de bisección calcula primero el punto medio:

$$
m=\frac{a+b}{2}
$$

El comportamiento del método depende de la posición de este punto medio respecto al valor:

$$
x=1
$$

* a) Si \(a+b<2\)

Entonces:

$$
\frac{a+b}{2}<1
$$

es decir:

$$
m<1
$$

El punto medio queda a la izquierda de \(x=1\).

En esta situación, el método de bisección conserva el subintervalo donde el cambio de signo ocurre alrededor de:

$$
x=0
$$

Por tanto, el método converge a:

$$
x=0
$$


* b) Si \(a+b>2\)

Entonces:

$$
\frac{a+b}{2}>1
$$

o equivalentemente:

$$
m>1
$$

El punto medio queda a la derecha de \(x=1\).

En consecuencia, el método conserva el subintervalo cuyo cambio de signo está alrededor de:

$$
x=2
$$

Por lo tanto, el método converge a:

$$
x=2
$$

* c) Si \(a+b=2\)

Entonces:

$$
\frac{a+b}{2}=1
$$

El primer punto medio es exactamente:

$$
m=1
$$

Evaluando la función:

$$
f(1)=\sin(\pi)=0
$$

Como el punto medio ya es una raíz exacta, el método termina inmediatamente.

Por tanto, el método converge a:

$$
x=1
$$
